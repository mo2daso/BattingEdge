```python
import os
import glob
import pandas as pd

print("--- DATASET SCANNER ---")

# 1. Define Paths
current_dir = os.getcwd()
if current_dir.endswith("notebooks"):
    PROJECT_ROOT = os.path.dirname(current_dir)
else:
    PROJECT_ROOT = current_dir

BASE_DIR = os.path.join(PROJECT_ROOT, 'data', 'videos_for_shot_classifier')
print(f"Scanning: {BASE_DIR}\n")

if not os.path.exists(BASE_DIR):
    print(f"❌ ERROR: Folder not found. Did you copy the files into data/videos_for_shot_classifier?")
else:
    # 2. Scan all subfolders
    splits = ['train', 'validation', 'test']
    all_files = []
    
    for split in splits:
        split_path = os.path.join(BASE_DIR, split)
        if not os.path.exists(split_path):
            print(f"⚠️ Warning: Split folder '{split}' not found.")
            continue
            
        # Use glob to find all video files recursively
        # ** means "any folder in between"
        video_files = glob.glob(os.path.join(split_path, '**', '*.avi'), recursive=True)
        video_files.extend(glob.glob(os.path.join(split_path, '**', '*.mp4'), recursive=True))
        
        for file in video_files:
            rel_path = os.path.relpath(file, BASE_DIR)
            shot_type = os.path.basename(os.path.dirname(file))
            all_files.append({
                'split': split,
                'shot_type': shot_type,
                'path': rel_path
            })

    # 3. Create a Report
    if len(all_files) > 0:
        df = pd.DataFrame(all_files)
        
        print(f"✅ SUCCESS! Found {len(df)} total video files.")
        
        print("\n--- 1. Report by Split ---")
        print(df['split'].value_counts())
        
        print("\n--- 2. Report by Shot Type (All Splits) ---")
        print(df['shot_type'].value_counts())
        
        print("\n--- 3. Example File Paths ---")
        print(df.head())
        
    else:
        print("❌ ERROR: No .avi or .mp4 files were found. Check your folder structure.")
```
