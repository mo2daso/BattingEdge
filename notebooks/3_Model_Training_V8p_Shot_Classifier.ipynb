````python
# File: notebooks/4_Train_V8p_Balanced.ipynb

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models, callbacks
from sklearn.preprocessing import LabelEncoder
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report, confusion_matrix
import joblib
import json
from pathlib import Path
from tqdm import tqdm
import matplotlib.pyplot as plt

# ============================================
# PATHS
# ============================================
ROOT = Path("..")
FEATURE_ROOT = ROOT / "data" / "features" / "dataset_v8p"
BACKEND_MODELS = ROOT / "backend" / "models"
BACKEND_MODELS.mkdir(parents=True, exist_ok=True)

CLASSES = ["drive", "pull", "cut", "sweep"]
SEQUENCE_LENGTH = 50
FEATURES_DIM = 103

# ============================================
# LOAD DATA
# ============================================
def load_split(split_name):
    """Load all .npy files for a split"""
    X_list, y_list = [], []
    
    for cls in CLASSES:
        cls_dir = FEATURE_ROOT / split_name / cls
        files = sorted(cls_dir.glob("*.npy"))
        
        print(f"Loading {split_name}/{cls}: {len(files)} files")
        
        for f in tqdm(files, desc=f"{cls}", leave=False):
            try:
                arr = np.load(f)
                
                # Validate shape
                if arr.shape != (SEQUENCE_LENGTH, FEATURES_DIM):
                    print(f"⚠️ Bad shape {arr.shape} in {f.name}, skipping")
                    continue
                
                # Check for NaN/Inf
                if np.isnan(arr).any() or np.isinf(arr).any():
                    print(f"⚠️ NaN/Inf in {f.name}, skipping")
                    continue
                
                X_list.append(arr)
                y_list.append(cls)
                
            except Exception as e:
                print(f"⚠️ Error loading {f.name}: {e}")
    
    X = np.array(X_list, dtype=np.float32)
    y = np.array(y_list)
    
    print(f"✅ {split_name}: {X.shape[0]} samples")
    return X, y

print("="*60)
print("LOADING DATASET")
print("="*60)

X_train, y_train = load_split("train")
X_val, y_val = load_split("val")
X_test, y_test = load_split("test")

# ============================================
# ENCODE LABELS
# ============================================
le = LabelEncoder()
le.fit(CLASSES)

y_train_int = le.transform(y_train)
y_val_int = le.transform(y_val)
y_test_int = le.transform(y_test)

# One-hot encode
num_classes = len(CLASSES)
y_train_oh = tf.keras.utils.to_categorical(y_train_int, num_classes)
y_val_oh = tf.keras.utils.to_categorical(y_val_int, num_classes)

# Save encoder
joblib.dump(le, BACKEND_MODELS / "shot_encoder_V8p.pkl")
print(f"✅ Saved encoder: {BACKEND_MODELS / 'shot_encoder_V8p.pkl'}")

# ============================================
# SCALING
# ============================================
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Fit on TRAIN data only (flatten to 2D for scaler)
N_train, T, D = X_train.shape
X_train_flat = X_train.reshape(N_train * T, D)
scaler.fit(X_train_flat)

# Transform all splits
def scale_data(X):
    N, T, D = X.shape
    X_flat = X.reshape(N * T, D)
    X_scaled = scaler.transform(X_flat)
    return X_scaled.reshape(N, T, D)

X_train = scale_data(X_train)
X_val = scale_data(X_val)
X_test = scale_data(X_test)

# Save scaler
joblib.dump(scaler, BACKEND_MODELS / "shot_scaler_V8p.pkl")
print(f"✅ Saved scaler: {BACKEND_MODELS / 'shot_scaler_V8p.pkl'}")

# ============================================
# CLASS DISTRIBUTION
# ============================================
from collections import Counter

print("\n" + "="*60)
print("CLASS DISTRIBUTION")
print("="*60)

for split_name, y_data in [("TRAIN", y_train), ("VAL", y_val), ("TEST", y_test)]:
    counts = Counter(y_data)
    print(f"\n{split_name}:")
    for cls in CLASSES:
        print(f"  {cls}: {counts[cls]}")

# ============================================
# CLASS WEIGHTS
# ============================================
class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.arange(num_classes),
    y=y_train_int
)
class_weight_dict = dict(enumerate(class_weights))

print("\n" + "="*60)
print("CLASS WEIGHTS")
print("="*60)
for i, cls in enumerate(CLASSES):
    print(f"{cls}: {class_weight_dict[i]:.3f}")

# ============================================
# MODEL ARCHITECTURE
# ============================================
def build_model():
    inputs = layers.Input(shape=(SEQUENCE_LENGTH, FEATURES_DIM))
    
    # Masking for variable-length sequences (if any)
    x = layers.Masking(mask_value=0.0)(inputs)
    
    # Bi-LSTM layers
    x = layers.Bidirectional(layers.LSTM(128, return_sequences=True))(x)
    x = layers.Dropout(0.3)(x)
    
    x = layers.Bidirectional(layers.LSTM(64, return_sequences=False))(x)
    x = layers.Dropout(0.3)(x)
    
    # Dense layers
    x = layers.Dense(64, activation='relu')(x)
    x = layers.Dropout(0.2)(x)
    
    # Output
    outputs = layers.Dense(num_classes, activation='softmax')(x)
    
    model = models.Model(inputs=inputs, outputs=outputs)
    
    # Compile
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    
    return model

model = build_model()
model.summary()

# ============================================
# CALLBACKS
# ============================================
checkpoint_path = BACKEND_MODELS / "shot_model_V8p_best.keras"

early_stop = callbacks.EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True,
    verbose=1
)

checkpoint = callbacks.ModelCheckpoint(
    str(checkpoint_path),
    monitor='val_accuracy',
    save_best_only=True,
    verbose=1
)

reduce_lr = callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=5,
    min_lr=1e-6,
    verbose=1
)

# ============================================
# TRAINING
# ============================================
print("\n" + "="*60)
print("TRAINING MODEL")
print("="*60)

EPOCHS = 60
BATCH_SIZE = 16

history = model.fit(
    X_train, y_train_oh,
    validation_data=(X_val, y_val_oh),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    class_weight=class_weight_dict,
    callbacks=[early_stop, checkpoint, reduce_lr],
    verbose=1
)

# Save final model as "brain"
model.save(BACKEND_MODELS / "shot_brain_V8p.keras")
print(f"✅ Saved brain model: {BACKEND_MODELS / 'shot_brain_V8p.keras'}")

# ============================================
# EVALUATION
# ============================================
print("\n" + "="*60)
print("EVALUATING ON TEST SET")
print("="*60)

# Load best model
best_model = tf.keras.models.load_model(str(checkpoint_path))

# Predict
y_pred_proba = best_model.predict(X_test, verbose=0)
y_pred = np.argmax(y_pred_proba, axis=1)

# Classification report
print("\nCLASSIFICATION REPORT:")
print(classification_report(y_test_int, y_pred, target_names=CLASSES, digits=3))

# Confusion matrix
cm = confusion_matrix(y_test_int, y_pred)
print("\nCONFUSION MATRIX:")
print("Rows = True, Cols = Predicted")
print(cm)

# Per-class accuracy
print("\nPER-CLASS ACCURACY:")
for i, cls in enumerate(CLASSES):
    correct = cm[i, i]
    total = cm[i, :].sum()
    acc = (correct / total * 100) if total > 0 else 0
    print(f"{cls}: {correct}/{total} = {acc:.1f}%")

# Overall accuracy
overall_acc = np.trace(cm) / np.sum(cm) * 100
print(f"\nOVERALL ACCURACY: {overall_acc:.2f}%")

# ============================================
# PLOT TRAINING HISTORY
# ============================================
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.title('Training Loss')
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Train Acc')
plt.plot(history.history['val_accuracy'], label='Val Acc')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.title('Training Accuracy')
plt.grid(True)

plt.tight_layout()
plt.savefig(BACKEND_MODELS / "training_history_V8p.png")
print(f"\n✅ Saved training plot: {BACKEND_MODELS / 'training_history_V8p.png'}")

plt.show()

# ============================================
# SAVE METADATA
# ============================================
metadata = {
    "classes": CLASSES,
    "sequence_length": SEQUENCE_LENGTH,
    "features_dim": FEATURES_DIM,
    "train_samples": len(X_train),
    "val_samples": len(X_val),
    "test_samples": len(X_test),
    "test_accuracy": float(overall_acc),
    "epochs_trained": len(history.history['loss'])
}

with open(BACKEND_MODELS / "model_V8p_metadata.json", "w") as f:
    json.dump(metadata, f, indent=2)

print(f"✅ Saved metadata: {BACKEND_MODELS / 'model_V8p_metadata.json'}")

print("\n" + "="*60)
print("TRAINING COMPLETE!")
print("="*60)
print("\nSaved artifacts:")
print(f"  - Model: {checkpoint_path}")
print(f"  - Brain: {BACKEND_MODELS / 'shot_brain_V8p.keras'}")
print(f"  - Encoder: {BACKEND_MODELS / 'shot_encoder_V8p.pkl'}")
print(f"  - Scaler: {BACKEND_MODELS / 'shot_scaler_V8p.pkl'}")
print(f"  - Metadata: {BACKEND_MODELS / 'model_V8p_metadata.json'}")
##```

##---

## **STEP 2: RUN TRAINING & REPORT RESULTS**

##**Run the notebook above**. It will output:
##```
##OVERALL ACCURACY: XX.XX%

##Per-class:
##- drive: XX%
##- pull: XX%
##- cut: XX%
##- sweep: XX%
````
