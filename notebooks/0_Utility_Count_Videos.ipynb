```python
import os

# Define your clean dataset path
# MAKE SURE THIS MATCHES WHERE YOU MOVED THE FILES
DATA_DIR = os.path.join(os.getcwd(), '..', 'data', 'dataset_v7_clean') 
# Or if you are still in the old structure:
# DATA_DIR = os.path.join(os.getcwd(), '..', 'data', 'videos_for_shot_classifier')

print(f"Scanning: {DATA_DIR}")

splits = ['train', 'val', 'test']
total_videos = 0

for split in splits:
    split_path = os.path.join(DATA_DIR, split)
    if not os.path.exists(split_path):
        continue
        
    print(f"\n--- {split.upper()} SET ---")
    split_total = 0
    
    # List all subfolders (shot types)
    shots = [d for d in os.listdir(split_path) if os.path.isdir(os.path.join(split_path, d))]
    
    for shot in shots:
        shot_path = os.path.join(split_path, shot)
        # Count files
        files = [f for f in os.listdir(shot_path) if f.endswith(('.mp4', '.avi'))]
        count = len(files)
        split_total += count
        print(f"  {shot}: {count}")
        
    print(f"  TOTAL {split}: {split_total}")
    total_videos += split_total

print(f"\n✅ GRAND TOTAL: {total_videos} videos")
```
