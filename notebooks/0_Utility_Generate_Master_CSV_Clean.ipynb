```python
import os
import glob
import pandas as pd
from pathlib import Path

print("--- GENERATING MASTER CSV FOR CLEAN DATASET ---")

# 1. SETUP PATHS
current_dir = Path(os.getcwd())
PROJECT_ROOT = current_dir.parent if current_dir.name == "notebooks" else current_dir

# This is your NEW clean dataset
VIDEO_DIR = PROJECT_ROOT / 'data' / 'dataset_v7_clean'
OUTPUT_CSV = PROJECT_ROOT / 'data' / 'master_labels.csv'

print(f"📂 Scanning: {VIDEO_DIR}")

# 2. SCAN FOLDERS
data_rows = []

# Find ALL video files recursively
video_files = list(VIDEO_DIR.rglob('*.mp4')) + list(VIDEO_DIR.rglob('*.avi'))

print(f"   Found {len(video_files)} videos.")

for video_path in video_files:
    # Structure: .../train/drive/drive_train_001.mp4
    # Parent folder is the Shot Type
    shot_type = video_path.parent.name
    split = video_path.parent.parent.name
    
    # Calculate relative path
    rel_path = video_path.relative_to(PROJECT_ROOT)
    
    data_rows.append({
        'filename': video_path.name,
        'shot_type': shot_type,
        'split': split,
        'video_path': str(rel_path),
        # YOUR JOB:
        'is_error': 0,          # 0=Good, 1=Bad
        'error_group': 'GOOD'   # GOOD, FOOT, UPPER
    })

# 3. SAVE CSV
df = pd.DataFrame(data_rows)
# Sort by shot_type so you can label them in batches
df = df.sort_values(by=['shot_type', 'split', 'filename'])

df.to_csv(OUTPUT_CSV, index=False)

print(f"✅ CSV Generated: {OUTPUT_CSV}")
print(f"📋 Total Rows: {len(df)}")
print("👉 Open 'master_labels.csv'. Filter by 'pull', watch 50 videos, label them. Repeat for 'drive'.")
```
