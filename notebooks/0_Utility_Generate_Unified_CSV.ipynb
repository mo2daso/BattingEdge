```python
import os
import glob
import pandas as pd
from pathlib import Path

print("--- GENERATING MASTER LABELING CSV (V7) ---")

# 1. SETUP PATHS
current_dir = Path(os.getcwd())
PROJECT_ROOT = current_dir.parent if current_dir.name == "notebooks" else current_dir

# This is your CLEAN, organized folder with 2,842 videos
VIDEO_DIR = PROJECT_ROOT / 'data' / 'dataset_v7_clean'
OUTPUT_CSV = PROJECT_ROOT / 'data' / 'master_labels.csv'

print(f"📂 Scanning: {VIDEO_DIR}")

# 2. SCAN FOLDERS
data_rows = []

# Recursive glob to find ALL video files in train/test/val
# We search for .mp4, .avi, .mov
extensions = ['*.mp4', '*.avi', '*.mov']
all_video_files = []
for ext in extensions:
    all_video_files.extend(list(VIDEO_DIR.rglob(ext)))

print(f"   Found {len(all_video_files)} videos.")

for video_path in all_video_files:
    # Path structure is: .../dataset_v7_clean/train/drive/drive_train_001.mp4
    # We can extract the SHOT TYPE from the parent folder name
    shot_type = video_path.parent.name # e.g., 'drive', 'pull'
    
    # Calculate relative path for the database
    rel_path = video_path.relative_to(PROJECT_ROOT)
    
    data_rows.append({
        'video_path': str(rel_path),
        'filename': video_path.name,
        'shot_type': shot_type,
        # COLUMNS FOR YOU TO FILL:
        'is_error': 0,      # Default to 0 (Good). Change to 1 if Bad.
        'error_group': 'GOOD' # Default to GOOD. Change to FOOT or UPPER.
    })

# 3. SAVE CSV
df = pd.DataFrame(data_rows)

# Sort to group similar shots together (makes labeling faster)
df = df.sort_values(by=['shot_type', 'filename'])

df.to_csv(OUTPUT_CSV, index=False)

print(f"✅ CSV Generated at: {OUTPUT_CSV}")
print(f"📋 Columns created: {list(df.columns)}")
print("👉 Open this file in Excel. It has defaults set to 'Good'. Only change the bad ones!")
```
