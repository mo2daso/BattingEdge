```python
import os
import glob
from pathlib import Path

print("--- STARTING DATASET CLEANUP & RENAME ---")

# ==========================================
# 1. CONFIGURATION
# ==========================================
current_dir = Path(os.getcwd())
PROJECT_ROOT = current_dir.parent if current_dir.name == "notebooks" else current_dir

# TARGET DIRECTORY (The Clean V7 Dataset)
DATA_DIR = PROJECT_ROOT / 'data' / 'dataset_v7_clean'
MIN_SIZE_BYTES = 2048 # 2 KB threshold

print(f"📂 Target Directory: {DATA_DIR}")
print(f"🗑️  Delete Threshold: < {MIN_SIZE_BYTES} bytes")

if not DATA_DIR.exists():
    raise ValueError(f"❌ Error: {DATA_DIR} does not exist!")

# ==========================================
# 2. CLEANUP (DELETE SMALL FILES)
# ==========================================
print("\n--- 1. DELETING CORRUPT FILES ---")
deleted_count = 0

# Find all video files recursively
all_files = list(DATA_DIR.rglob('*.mp4')) + list(DATA_DIR.rglob('*.avi')) + list(DATA_DIR.rglob('*.mov'))

for file_path in all_files:
    try:
        size = file_path.stat().st_size
        if size < MIN_SIZE_BYTES:
            print(f"   🗑️ Deleting {file_path.name} ({size} bytes)")
            file_path.unlink() # DELETE COMMAND
            deleted_count += 1
    except Exception as e:
        print(f"   ⚠️ Error accessing {file_path.name}: {e}")

print(f"✅ Cleanup Complete. Deleted {deleted_count} corrupt files.")

# ==========================================
# 3. RENAME & RE-INDEX
# ==========================================
print("\n--- 2. RENAMING & RE-INDEXING ---")
renamed_count = 0

splits = ['train', 'val', 'test']

for split in splits:
    split_dir = DATA_DIR / split
    if not split_dir.exists(): continue
    
    # Get shot folders (drive, pull, etc.)
    shot_folders = [d for d in split_dir.iterdir() if d.is_dir()]
    
    for shot_dir in shot_folders:
        shot_name = shot_dir.name
        
        # Get all valid videos remaining
        videos = sorted(list(shot_dir.glob('*.mp4')) + list(shot_dir.glob('*.avi')))
        
        if not videos: continue
        
        print(f"   Re-indexing {split}/{shot_name} ({len(videos)} files)...")
        
        for index, video in enumerate(videos, 1):
            # Format: drive_train_001.mp4
            new_name = f"{shot_name}_{split}_{index:03d}{video.suffix}"
            new_path = shot_dir / new_name
            
            # Only rename if name is different to avoid errors
            if video.name != new_name:
                try:
                    video.rename(new_path)
                    renamed_count += 1
                except Exception as e:
                    print(f"      ⚠️ Could not rename {video.name}: {e}")

print(f"✅ Renaming Complete. Renamed {renamed_count} files.")

# ==========================================
# 4. FINAL COUNT
# ==========================================
print("\n--- 3. FINAL DATASET REPORT ---")
total_videos = 0

for split in splits:
    print(f"\n[{split.upper()}]")
    split_dir = DATA_DIR / split
    if not split_dir.exists(): continue
    
    shot_folders = [d for d in split_dir.iterdir() if d.is_dir()]
    for shot_dir in shot_folders:
        count = len(list(shot_dir.glob('*.*'))) # Count all files
        print(f"   - {shot_dir.name}: {count}")
        total_videos += count

print("\n" + "="*40)
print(f"🚀 TOTAL CLEAN DATASET SIZE: {total_videos} VIDEOS")
print("="*40)
```
